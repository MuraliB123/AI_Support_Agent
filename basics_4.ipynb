```python
from typing import List, Dict, TypedDict
from langgraph.graph import StateGraph, START, END
import random
```

```python
class AgentState(TypedDict):
    name: str
    number: List[int]
    counter: int
```

```python
def print_node(state: AgentState) -> AgentState:
    """ print message """
    state['name'] = f'Hi {state['name']} How are you doing ?'
    state['counter'] = 0
    return state
```

```python
def random_node(state: AgentState) -> AgentState:
    """ do work and increment counter """
    state['number'].append(random.randint(0,10))
    state['counter'] += 1
    return state
```

```python
def condition(state: AgentState) -> str:
    """ condition for edge """
    if state['counter'] < 5:
        return "loop"
    else:
        return "exit"
```

```python
graph = StateGraph(AgentState)
graph.add_node("greeting",print_node)
graph.add_node("random",random_node)
graph.add_edge("greeting","random")
graph.add_conditional_edges("random",
                           condition,
                           {
                               "loop": "random",
                               "exit": END
                           }
                          )
graph.set_entry_point("greeting")
app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```
