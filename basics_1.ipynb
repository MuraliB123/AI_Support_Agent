```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    result: str
```

```python
def process_values(state: AgentState) -> AgentState:
    """ Hmm this function just perform operations in integer list"""
    print(state)
    state['result'] = f"Hi {state['name']}, your answer is {sum(state['values'])}"
    print(state)
    return state
```

```python
graph = StateGraph(AgentState)
```

```python
graph.add_node("sum_ops",process_values)
```

```python
graph.set_entry_point("sum_ops")
graph.set_finish_point("sum_ops")
```

```python
agent = graph.compile()
```

```python
result = agent.invoke({'values':[1,2,3,4,5],'name':'Alan stark'})
```
