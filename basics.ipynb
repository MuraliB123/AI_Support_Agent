```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    message : str
```

```python
def greeting_node(state: AgentState) -> AgentState:
    """ Simple node for printing greeting message"""
    state['message'] = "Hey " + state['message'] + " How are you doing!"
    return state
```

```python
graph = StateGraph(AgentState)
```

```python
graph.add_node("welcome",greeting_node)
graph.set_entry_point("welcome")
graph.set_finish_point("welcome")
```

```python
agent = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(agent.get_graph().draw_mermaid_png()))
```

```python
result = agent.invoke({"message": "Stark"})
```
